### Q2

Desenvolva uma função em Python chamada ```tem_elementos_comuns()``` que recebe duas listas como parâmetros e retorna True se houver algum elemento comum entre as duas listas, e False caso contrário. Utilize sets para resolver essa tarefa.

````python
# Implemente aqui a função tem_elementos_comuns()
def tem_elementos_comuns(lista1, lista2):
    # Converte as listas em sets e verifica se há interseção
    return bool(set(lista1) & set(lista2))
````

````python
# Execute esta célula para testar a função da Q2.

# Teste da função
lista1 = [1, 2, 3, 4]
lista2 = [3, 4, 5, 6, 7]
resultado = tem_elementos_comuns(lista1, lista2)
print(resultado)  # Saída: True

# Mais casos de teste
print(tem_elementos_comuns([1, 2, 3], [4, 5, 6]))  # False
print(tem_elementos_comuns([1, 2], [2, 3]))        # True
print(tem_elementos_comuns([], [1, 2, 3]))         # False
print(tem_elementos_comuns([1, 1, 2], [2, 2]))     # True
````

### Q4.
Considere duas listas que são duplicatas uma da outra, exceto por um elemento, exemplo:
```
A = [1, 4, 5, 7, 9]
B = [4, 5, 7, 9]
```

Encontre e apresente o elemento diferente, informando também qual das listas está desfalcada. Para o exemplo acima, a saída esperada é
```
O elemento 1 está faltando na segunda lista
```

````python
def encontrar_elemento_faltando(lista1, lista2):
    # Verifica qual lista é maior para determinar a ordem de comparação
    if len(lista1) > len(lista2):
        lista_maior = lista1
        lista_menor = lista2
        lista_desfalcada = "segunda lista"
    else:
        lista_maior = lista2
        lista_menor = lista1
        lista_desfalcada = "primeira lista"

    # Encontra o elemento que está na lista maior mas não na menor
    for elemento in lista_maior:
        if elemento not in lista_menor:
            return f"O elemento {elemento} está faltando na {lista_desfalcada}."

    return "Nenhuma diferença encontrada."

# Exemplo fornecido
A = [1, 4, 5, 7, 9]
B = [4, 5, 7, 9]

# Testando a função
resultado = encontrar_elemento_faltando(A, B)
print(resultado)
````

### Q5.

Panagramas são frases que contém todas as letras do alfabeto pelo menos uma vez. Crie uma função ```checa_panagrama``` que recebe uma string e retorna ```True``` caso seja um panagrama ou ```False```caso contrário. Use seu conhecimento de sets para solucionar essa questão.

Ex:
```
Entrada: "The quick brown fox jumps over the lazy dog"
Saída: É um panagrama
```

```
Entrada: "Python é uma linguagem de programação"
Saída: Não é um panagrama
```

````python
def checa_panagrama(frase):
    # Criar um set com todas as letras do alfabeto (a-z)
    alfabeto = set('abcdefghijklmnopqrstuvwxyz')

    # Converter a frase para minúsculas e criar um seta apenas com letras
    letras_frase = set(char for char in frase.lower() if char.isalpha())

    # Verificar se todas as letras do alfabeto estão presentes na frase
    if alfabeto.issubset(letras_frase):
        return "É um pangrama"
    return "Não é um pangrama"

# Testes
print(checa_panagrama("The quick brown fox jumps over the lazy dog"))
print(checa_panagrama("Python é uma linguagem de programação"))
````
